```python
import pandas as pd
import numpy as np

def calculate_unit_economics(
    ua: int, 
    cr1: float, 
    aov: float, 
    lifespan_months: float, 
    ad_budget: float, 
    gross_margin: float = 1.0
) -> dict:
    """
    Calculates key Unit Economics metrics for subscription business model.
    """
    buyers = round(ua * cr1)
    
    # LTV = AOV * Lifespan * Gross Margin
    ltv = round(aov * lifespan_months * gross_margin, 2)
    
    # CAC = Ad Budget / Total Buyers
    cac = round(ad_budget / buyers, 2) if buyers > 0 else 0
    
    # LTV / CAC Ratio
    ltv_to_cac = round(ltv / cac, 2) if cac > 0 else 0
    
    # Unit ROMI = ((LTV - CAC) / CAC) * 100
    romi = round(((ltv - cac) / cac) * 100, 1) if cac > 0 else 0
    
    # Total Financials
    total_revenue = round(buyers * ltv, 2)
    net_profit = round(total_revenue - ad_budget, 2)

    return {
        "Acquired Users (UA)": ua,
        "Paying Customers": buyers,
        "AOV ($)": aov,
        "Lifespan (Months)": lifespan_months,
        "LTV ($)": ltv,
        "CAC ($)": cac,
        "LTV / CAC Ratio": f"{ltv_to_cac}x",
        "Unit ROMI (%)": f"{romi}%",
        "Total Revenue ($)": total_revenue,
        "Net Profit ($)": net_profit
    }

# --- Baseline Business Performance (Pre-Experiment) ---
params = {
    "ua": 100000,
    "cr1": 0.015,          # 1.5% trial-to-paid conversion
    "aov": 9.99,           # Monthly subscription price
    "lifespan_months": 4.2,# Average user lifespan in months
    "ad_budget": 45000,    # Total marketing spend
    "gross_margin": 1.0    # Assumed 100% gross margin
}

metrics = calculate_unit_economics(**params)

# Display results in structured DataFrame
df_results = pd.DataFrame(list(metrics.items()), columns=["Metric", "Value"])
print("=== Baseline Unit Economics Model ===")
print(df_results.to_string(index=False))
```
