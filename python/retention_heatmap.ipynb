```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Data Assembly (Weekly Cohorts)
data = {
    'Cohort': [
        'Week 20 (n=1,500)', 
        'Week 21 (n=1,620)', 
        'Week 22 (n=1,480)', 
        'Week 23 (n=1,710)', 
        'Week 24 (n=1,550)'
    ],
    'Week 0': [100.0, 100.0, 100.0, 100.0, 100.0],
    'Week 1': [35.4, 38.1, 32.0, 39.5, 41.2],
    'Week 2': [22.1, 24.5, 18.2, 25.1, np.nan],
    'Week 3': [15.8, 17.0, 11.5, np.nan, np.nan],
    'Week 4': [12.0, 11.2, np.nan, np.nan, np.nan]
}

cohort_pivot = pd.DataFrame(data).set_index('Cohort')

# 2. Style Setup
plt.figure(figsize=(10, 5), dpi=300)
sns.set_theme(style="white")

# 3. Heatmap Construction (vmax capped at 50)
ax = sns.heatmap(
    cohort_pivot, 
    annot=True, 
    fmt=".1f", 
    cmap="Blues", 
    vmin=0, 
    vmax=50,  # Cap color bar range at 50%
    linewidths=1, 
    linecolor='white',
    cbar_kws={'label': '% Retention'},
    annot_kws={"size": 11, "weight": "bold"}
)

# 4. Title & Axis Customization
plt.title('Weekly Retention Rate (%) by User Cohorts', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('User Lifecycle Stage (Weeks)', fontsize=11, fontweight='bold', labelpad=10)
plt.ylabel('Registration Cohort', fontsize=11, fontweight='bold', labelpad=10)

# 5. Add % Sign to Cell Values
for text in ax.texts:
    val = text.get_text()
    if val != 'nan':
        text.set_text(f"{val}%")

plt.tight_layout()

# 6. Save & Display
plt.savefig('retention_heatmap_weekly.png', dpi=300, bbox_inches='tight')
plt.show()
```
